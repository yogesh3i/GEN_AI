```python
# Settign up the embeddign models
from langchain_community.embeddings import HuggingFaceEmbeddings

# set the model
model_name = "sentence-transformers/all-mpnet-base-v2"
# model_kwargs = {'device': 'cpu'}
encode_kwargs = {'normalize_embeddings': False}

embeddings = HuggingFaceEmbeddings(
    model_name=model_name

)
```

```python
# pdf loader
from langchain_community.document_loaders import PyPDFLoader
loader = PyPDFLoader("/content/RAG_paper.pdf",extract_images=True)
pages = loader.load()
```

```python
pages
```

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size = 1000,
    chunk_overlap  = 20,

)

# split it
docs = text_splitter.split_documents(pages)
```

```python
vector_db = Weaviate.from_documents(
    docs,embeddings,client=client,by_text=False
)
```

```python
vector_db.similarity_search("What is RAG ?",k=3)[1].page_content
```

```python
from langchain.prompts import ChatPromptTemplate

template = """You are assistance for a question answering task.
Use the following piece of retrieved context to answer the question.
If you dont know the answer please , Id dont about this.
Use the max 10 sentences and keep the asnwer concise.
Question:{question}
context:{context}
Answer:
"""
```

```python
prompt = ChatPromptTemplate.from_template(template)
```

```python
from google.colab import userdata
hugging_face_api_key = userdata.get("hugging_face_key")
```

```python
from langchain_community.llms import HuggingFaceHub

model = HuggingFaceHub(
    huggingfacehub_api_token=hugging_face_api_key,
    repo_id = "mistralai/Mistral-7B-v0.1",
    model_kwargs={"temperature":0.7,"max_length":180}
)
```

```python
# generate the asnwers
from langchain.schema.runnable import RunnablePassthrough
from langchain.schema.output_parser import StrOutputParser
```

```python
# create output parser
output_parsers = StrOutputParser()
retriver = vector_db.as_retriever()
```

```python
# rag chain

rag_chain = (
    {"contect":retriever, "question":RunnablePassthrough()}
    | prompt
    | model
    | StrOutputParser()
)
```
