### **2. Using StructuredTool**

```python
from langchain.tools import StructuredTool
from pydantic import BaseModel, Field
```

```python
from langchain.tools import StructuredTool
from pydantic import BaseModel, Field

# ✅ 1. Define Pydantic input model
class MultiplyInput(BaseModel):
    a: int = Field(..., description="First number")
    b: int = Field(..., description="Second number")

# ✅ 2. Define function
def MultiplyFunc(a: int, b: int) -> int:
    return a * b

# ✅ 3. Create the tool
multiply_tool = StructuredTool.from_function(
    func=MultiplyFunc,
    name="Multiply",
    description="Multiply two numbers",
    args_schema=MultiplyInput
)

# ✅ 4. Use the tool
result = multiply_tool.invoke({"a": 6, "b": 7})
print(result)  # Output: 42
```

```python
# Defien the pydeantic class

class Add_two_num(BaseModel):
  a : int = Field(...,dexcription=True)
  b: int = Field(...,description=True)
```

```python
# define a function
def Addition(a:int, b:int) ->int:
  return f"The summation is {a+b}"
```

```python
# structure the tool

additoin_tool = StructuredTool.from_function(
    func=Addition,
    name="Additon",
    description = "Add the given two numbers",
    args_schema=Add_two_num
)
```

```python
# Calling the tool
additoin_tool.invoke({"a":5,"b":7})
```

### **Base Tool**

The main structure of the tool on which all the tools of langchain are built.

```python
from langchain.tools import BaseTool
from typing import Type

# define the arg schema
class substract(BaseModel):
  a:int =Field(...,description="Enter first number")
  b:int =Field(...,description="Enter second number")

# no need to define the function now just define a class which will be child class of the
class Sub_tool(BaseTool):
  name: str="Substract"
  description: str="Substract two numbers"
  args_schema: Type[BaseModel]= substract

  def _run(self,a:int,b:int) -> int:
    return a-b
```

```python
# call the tool
sub = Sub_tool()
sub.invoke({"a":9,"b":10})
```

```python
#let's try building another on using BaseTool

#1- define the pydantic class

class div(BaseModel):
  a:int =Field(...,description="First number")
  b:int =Field(...,description="Second number")


#2- Define the toll as per basetool

class div_2_number(BaseTool):
  name:str ="Division"
  description:str ="Divide thetwo numbers"
  args_schema: Type[BaseModel]=div

  def _run(self,a:int,b:int) -> float:
    return f"The division is {a/b}"


```

```python
# call the tool
div_toll = div_2_number()
div_toll.invoke({"a":9,"b":3})
```

### **ToolKit**

A packet of multiplte tools togather.

```python
# tool-1
from langchain.tools import tool

# tool-1
@tool
def addition(a:int,b:int) -> int:
  """Add the two number"""
  return f"The addition is {a+b}"

# tool-2
@tool
def multiply(a:int,b:int) ->int:
  """Multiply two number"""
  return f"The multi id {a*b}"


# building the toolkit
class math_toolkit:
  def get_tools(self):
    return [addition,multiply]

# calling the toolkit
mathy= math_toolkit()

tools = mathy.get_tools()
```

```python
for tool in tools:
  print(tool.name,"->",tool.description)
```

## **Building Currency Exchange Tool**

```python
@tool
def currency_exchange(base_currency:str,target_currency:str) -> float:
  """This function fetches the currency conversion factor between a given base currency and a target currency"""

  url = f"https://v6.exchangerate-api.com/v6/26c1560949e060e9c50b84e5/pair/{base_currency}/{target_currency}"

  response = requests.get(url)
  return response.json()


# now the conversion factor is there build tool to make the conversion
@tool
def convert(base_currency_value:int,conversion_rate:float) ->float:
  """given a currency conversion rate this function calculates the target currency value from a given base currency value"""
  return base_currency_value*conversion_rate
```

```python
# try toll is working or not
currency_exchange.invoke({"base_currency":'USD',"target_currency":'INR'})
```

```python
convert({"base_currency_value":10,"conversion_rate":85.8681})
```

```python
# now bind this tools with llm
llm_with_tools = llm.bind_tools([currency_exchange,convert])
```

```python
print(llm_with_tools)
```

```python
message =[HumanMessage("What is thec onversion rate for USD to INR and convert the 10 USD to INR")]
```

```python
# invoke the llm with this message
ai_message = llm_with_tools.invoke(message)
```

```python
message.append(ai_message)
```

```python
import json

# impliment the condition when to call whcih tool
for tool_call in ai_message.tool_calls:
  if tool_call['name']=="currency_exchange":
    tool_message1 = currency_exchange.invoke(tool_call)
    # fetch this conversion rate
    conversion_rate = json.loads(tool_message1.content)['conversion_rate']
    # append the tool_message1 to the message
    message.append(tool_message1)

    # second tool
    if tool_call['name']=='convert':
      tool_message2 = convert.invoke(tool_call)
      message.append(tool_message2)
```
