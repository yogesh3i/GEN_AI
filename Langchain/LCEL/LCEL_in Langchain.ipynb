##### What are the runnables ?

##### Ans:
          Runnables are self contain unit of work, it can executed individually or can be execuated in combination whenever the complex operation is required.

##### 1) RunnableParallel():

#### Defination & Working:
          **RunnablePassThrough() is a class in the Langchian used to get the input data and pass it directly to the output without making any chanegs into it.Helpful in debugging and chaining the pipeline**

```python
from langchain_community.document_loaders import TextLoader,DirectoryLoader,PyPDFLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import Chroma

# reading the data
loader = DirectoryLoader("/content/data",loader_cls=PyPDFLoader)
docs = loader.load()

# perform chunking
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000,chunk_overlap=200)
new_docs = text_splitter.split_documents(docs)

doc_string = [doc.page_content for doc in new_docs]

# stroring embeddings
db = Chroma.from_documents(new_docs,embeddings)

# retriver
retriver = db.as_retriever(search_kwargs={"k":4})
```

```python
# promt
template = """Answer the question based on the following context :
{context}

Question:{question}
"""

prompt = PromptTemplate.from_template(template)
```

```python
# retrieval chain

retriever_chain = RunnableParallel({"context":retriver,"question":RunnablePassthrough()}) | prompt | llm | StrOutputParser()
```

```python
retriever_chain.invoke("Can you explain the architecture of Yolo ?")
```

```python
# Lets find out the difference between sync and async
Question = "According to you please summerrize the document given to youin 200 words"
import time
start = time.time()
result = retriever_chain.invoke(Question)
print(result)
print("Time taken is:", time.time()-start)
```

```python
# Lets find out the difference between sync and async
Question = "According to you please summerrize the document given to youin 200 words"
import time
start = time.time()
result =  retriever_chain.ainvoke(Question)
print("Time taken is:", time.time()-start)
```

```python
# sync batch
start = time.time()

batch_output = retriever_chain.batch(["WHAT IS YOLO","hOW DOES IT WORK"])
print("time taken:", time.time()-start)
```

```python
# Chaining with itemgetter

template = """
Answer the question based on the following context:
{context}

Question: {question}
Answer the question in language: {language}
"""

prompt = PromptTemplate.from_template(template)
```

```python
retriever_chain = (
RunnableParallel({"context":itemgetter("question") | retriver,
                                    "question":itemgetter("question"),
                                    "language":itemgetter("language")
                                    })
                                    | prompt
                                    | llm
                                    |StrOutputParser()

)
```

```python
retriever_chain.invoke({"question":"What is yolo ?","language":"Hindi"})
```

```python
# Streaming th eout put for best user experiance

template = """Answer the question in the {language}
            Question:{question}"""

prompt = PromptTemplate.from_template(template)

chain = prompt | llm
```

```python
# Streaming th eout put for best user experiance

template = """Answer the question in the {language}
            Question:{question}"""

prompt = PromptTemplate.from_template(template)

chain = prompt | llm


for s in chain.stream({"language":"Hindi","question":"Summerize the yolo in 500 words"}):
  print(s.content,end=" ")
```
