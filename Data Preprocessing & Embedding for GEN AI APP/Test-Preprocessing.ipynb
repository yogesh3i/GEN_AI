**Punctuation handling**

```python
import string,time
string.punctuation
```

```python
exclude = string.punctuation
exclude
```

```python
def remove_punc(text):
    for char in exclude:
        text = text.replace(char,'')
    return text 
```

```python
text = 'string.With. Punctuation?'
```

```python
# let's calculate the time as well to remove the punctuation 

start = time.time()
print(remove_punc(text))
time1 = time.time()-start
print(time1*50000)
```

```python
# Another function to remove punctuations 
def remove_punc1(text):
    return text.translate(str.maketrans('','',exclude))
```

**chat_conversion handle**

```python
chat_words = {
    'AFAIK':'As Far As I Know',
    'AFK':'Away From Keyboard',
    'ASAP':'As Soon As Possible'
}

{
    "FYI": "For Your Information",
    "ASAP": "As Soon As Possible",
    "BRB": "Be Right Back",
    "BTW": "By The Way",
    "OMG": "Oh My God",
    "IMO": "In My Opinion",
    "LOL": "Laugh Out Loud",
    "TTYL": "Talk To You Later",
    "GTG": "Got To Go",
    "TTYT": "Talk To You Tomorrow",
    "IDK": "I Don't Know",
    "TMI": "Too Much Information",
    "IMHO": "In My Humble Opinion",
    "ICYMI": "In Case You Missed It",
    "AFAIK": "As Far As I Know",
    "BTW": "By The Way",
    "FAQ": "Frequently Asked Questions",
    "TGIF": "Thank God It's Friday",
    "FYA": "For Your Action",
    "ICYMI": "In Case You Missed It",
}    
```

```python
def chat_conversion(text):
    new_text = []
    for w in text.split():
        if w.upper() in chat_words:
            new_text.append(chat_words[w.upper()])
        else:
            new_text.append(w)
    return " ".join(new_text)
```

**stopwords**

- Why removel of the stop is necessary and why we don't need the stopwords in the sentence for sentiment analysis ??

Ex. I have below sentence to do the sentiment analysis. 

This movie is awesome i loved it. 

only the below words are required to get the sentiment of the this sentence. 

movie awesome loved -----> this shows the positive sentiment. 

Rest all words are stopwords they are just increasing the dimensionality hence need to remove.

```python
from nltk.corpus import stopwords 
import nltk 
nltk.download('stopwords')
```

```python
stopwords.words('english')
```

```python
len(stopwords.words('english'))
```

```python
def remove_stopwords(text):
    new_text = []

    for word in text.split():
        if word in stopwords.words('english'):
            new_text.append('')
        else:
            new_text.append(word)
    x = new_text[:]
    new_text.clear()
    return " ".join(x)
```

```python
remove_stopwords('probably my all-time favorite movie, a story of selflessness, sacrifice and dedication to a noble cause, but it\'s not preachy or boring. it just never gets old, despite my having seen it some 15 or more times')
```

```python
df['review'].apply(remove_stopwords)
```

**Remove Emoji**

```python
import re 
def remove_emoji(text):
    emoji_pattern =re.compile("["
                           u"\U0001F600-\U0001F64F"  # emoticons
                           u"\U0001F300-\U0001F5FF"  # symbols & pictographs
                           u"\U0001F680-\U0001F6FF"  # transport & map symbols
                           u"\U0001F1E0-\U0001F1FF"  # flags (iOS)
                           u"\U00002702-\U000027B0"
                           u"\U000024C2-\U0001F251"
                           "]+", flags=re.UNICODE)

    return emoji_pattern.sub(r'',text)
```

```python
remove_emoji("Loved the movie. It was 😘😘")
```

```python
remove_emoji("Lmao 😂😂")
```

**Lemmatization**

NOTE: Stemming & lamatization are same to retrieve root words but lamatization is worked good. Lamatization is slow & stemming is fast

```python
import nltk 
from nltk.stem import WordNetLemmatizer 
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')
wordnet_lemmitizer = WordNetLemmatizer()

sentence = "He was running and eating at same time. He has bad habit of swimming after playing long hours in the Sun."
punctuations="?:!.,;"
sentence_words =nltk.word_tokenize(sentence)

for word in sentence_words:
    if word in punctuations:
        sentence_words.remove(word)

sentence_words 
print("{0:20}{1:20}".format('word','Lemma'))
for word in sentence_words:
    print ("{0:20}{1:20}".format(word,wordnet_lemmitizer.lemmatize(word,pos='v')))
```
